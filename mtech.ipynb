```python
# Q1. Sum of digits of a number (input from user)
def sum_of_digits(num):
    if num < 0:
        num = -num
    total = 0
    while num > 0:
        total += num % 10
        num //= 10
    return total

n = int(input("Enter a number: "))
print("Sum of digits of", n, "is", sum_of_digits(n))
```

```python
# Q2. Armstrong number of order n
def power(base, exp):
    result = 1
    for _ in range(exp):
        result *= base
    return result

def is_armstrong(number, order):
    temp = number
    total = 0
    while temp > 0:
        digit = temp % 10
        total += power(digit, order)
        temp //= 10
    return total == number

number = int(input("Enter a number: "))
order = int(input("Enter the order n: "))
if is_armstrong(number, order):
    print(number, "is an Armstrong number of order", order)
else:
    print(number, "is NOT an Armstrong number of order", order)
```

```python
# Q3. Palindrome string check
def is_palindrome(s):
    left = 0
    right = len(s) - 1
    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True

s = input("Enter a string: ")
if is_palindrome(s):
    print('"' + s + '" is a palindrome')
else:
    print('"' + s + '" is not a palindrome')
```

```python
# Q4. Print all elements of an array and their frequencies
def print_frequencies(arr):
    freq = {}
    for x in arr:
        if x in freq:
            freq[x] += 1
        else:
            freq[x] = 1
    for key in freq:
        print(key, "->", freq[key])

arr = [int(x) for x in input("Enter array elements separated by space: ").split()]
print_frequencies(arr)
```

```python
# Q5. Print all prime factors of n  (12 -> 2 2 3)
def prime_factors(n):
    factors = []
    d = 2
    while d * d <= n:
        while n % d == 0:
            factors.append(d)
            n //= d
        d += 1
    if n > 1:
        factors.append(n)
    return factors

n = int(input("Enter a number: "))
for f in prime_factors(n):
    print(f, end=" ")
print()
```

```python
# Q6. Binomial coefficient nCr = n! / (r! * (n-r)!)
def factorial(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

def binomial(n, r):
    if r < 0 or r > n:
        return 0
    return factorial(n) // (factorial(r) * factorial(n - r))

n = int(input("Enter n: "))
r = int(input("Enter r: "))
print("C(" + str(n) + ", " + str(r) + ") =", binomial(n, r))
```

```python
# Q7. Linear and binary search on a sorted array
def linear_search(arr, x):
    for i in range(len(arr)):
        if arr[i] == x:
            return i
    return -1

def binary_search(arr, x):
    low = 0
    high = len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == x:
            return mid
        elif arr[mid] < x:
            low = mid + 1
        else:
            high = mid - 1
    return -1

arr = [int(v) for v in input("Enter SORTED array elements separated by space: ").split()]
x = int(input("Enter element to search: "))
print("Linear search index:", linear_search(arr, x))
print("Binary search index:", binary_search(arr, x))
```

```python
# Q8. Word frequency from a text file + 3 most frequent words
def clean_word(word):
    result = ""
    for ch in word.lower():
        if ("a" <= ch <= "z") or ("0" <= ch <= "9") or ch == "'":
            result += ch
    return result

def word_frequencies(filename):
    freq = {}
    f = open(filename, "r")
    for line in f:
        for raw in line.split():
            w = clean_word(raw)
            if w != "":
                if w in freq:
                    freq[w] += 1
                else:
                    freq[w] = 1
    f.close()
    return freq

def top_n_words(freq, n):
    items = []
    for w in freq:
        items.append((w, freq[w]))
    top = []
    for _ in range(n):
        if len(items) == 0:
            break
        best = 0
        for i in range(1, len(items)):
            if items[i][1] > items[best][1]:
                best = i
        top.append(items[best])
        items.pop(best)
    return top

filename = input("Enter the text file name: ")
try:
    freq = word_frequencies(filename)
    print("\nWord frequencies:")
    for w in freq:
        print(w, ":", freq[w])
    print("\n3 most frequent words:")
    for w, c in top_n_words(freq, 3):
        print(w, ":", c)
except FileNotFoundError:
    print("File not found.")
```

```python
# Q9. Hospital Patient Record System
patients = [
    {"name": "Amit",  "age": 45, "disease": "Diabetes",      "bill": 12000},
    {"name": "Sara",  "age": 30, "disease": "Flu",           "bill": 3000},
    {"name": "John",  "age": 52, "disease": "Diabetes",      "bill": 25000},
    {"name": "Priya", "age": 28, "disease": "Malaria",       "bill": 8000},
    {"name": "Sara",  "age": 30, "disease": "Flu",           "bill": 3000},   # duplicate
    {"name": "Ravi",  "age": 60, "disease": "Heart Disease", "bill": 90000},
    {"name": "Neha",  "age": 35, "disease": "Flu",           "bill": 4000},
]

def remove_duplicates(records):
    unique = []
    for rec in records:
        duplicate = False
        for u in unique:
            if (u["name"] == rec["name"] and u["age"] == rec["age"]
                    and u["disease"] == rec["disease"] and u["bill"] == rec["bill"]):
                duplicate = True
                break
        if not duplicate:
            unique.append(rec)
    return unique

def patients_per_disease(records):
    count = {}
    for rec in records:
        d = rec["disease"]
        count[d] = count.get(d, 0) + 1
    return count

def highest_bill_patient(records):
    best = records[0]
    for rec in records:
        if rec["bill"] > best["bill"]:
            best = rec
    return best

def average_bill(records):
    total = 0
    for rec in records:
        total += rec["bill"]
    return total / len(records)

print("Records before removing duplicates:", len(patients))
patients = remove_duplicates(patients)
print("Records after removing duplicates :", len(patients))

print("\nNumber of patients affected by each disease:")
counts = patients_per_disease(patients)
for d in counts:
    print(" ", d, ":", counts[d])

top = highest_bill_patient(patients)
print("\nPatient with highest bill:", top["name"], "- Rs.", top["bill"])

avg = average_bill(patients)
print("\nAverage bill: Rs.", round(avg, 2))
print("Patients whose bill exceeds the average:")
for rec in patients:
    if rec["bill"] > avg:
        print(" ", rec["name"], "(", rec["disease"], ") - Rs.", rec["bill"])
```

```python
# Online Food Delivery Analysis System
# Each order: order_id, restaurant, items [(item name, quantity, price)]
orders = [
    {"order_id": 1, "restaurant": "Pizza Hub",
     "items": [("Margherita Pizza", 2, 250), ("Garlic Bread", 1, 120)]},
    {"order_id": 2, "restaurant": "Biryani House",
     "items": [("Chicken Biryani", 3, 300)]},
    {"order_id": 3, "restaurant": "Pizza Hub",
     "items": [("Margherita Pizza", 1, 250), ("Coke", 2, 50)]},
    {"order_id": 2, "restaurant": "Biryani House",          # duplicate of order 2
     "items": [("Chicken Biryani", 3, 300)]},
    {"order_id": 4, "restaurant": "Burger Point",
     "items": [("Veg Burger", 2, 100), ("Fries", 2, 80)]},
    {"order_id": 5, "restaurant": "Biryani House",
     "items": [("Chicken Biryani", 1, 300), ("Raita", 1, 40)]},
]

def remove_duplicate_orders(order_list):
    unique = []
    for o in order_list:
        duplicate = False
        for u in unique:
            if (u["order_id"] == o["order_id"] and u["restaurant"] == o["restaurant"]
                    and u["items"] == o["items"]):
                duplicate = True
                break
        if not duplicate:
            unique.append(o)
    return unique

def revenue_by_restaurant(order_list):
    revenue = {}
    for o in order_list:
        order_total = 0
        for name, qty, price in o["items"]:
            order_total += qty * price
        r = o["restaurant"]
        revenue[r] = revenue.get(r, 0) + order_total
    return revenue

def most_ordered_item(order_list):
    qty_count = {}
    for o in order_list:
        for name, qty, price in o["items"]:
            qty_count[name] = qty_count.get(name, 0) + qty
    best_item, best_qty = None, 0
    for name in qty_count:
        if qty_count[name] > best_qty:
            best_item, best_qty = name, qty_count[name]
    return best_item, best_qty

def top_restaurant(revenue):
    best, best_rev = None, -1
    for r in revenue:
        if revenue[r] > best_rev:
            best, best_rev = r, revenue[r]
    return best, best_rev

print("Orders before removing duplicates:", len(orders))
orders = remove_duplicate_orders(orders)
print("Orders after removing duplicates :", len(orders))

print("\nTotal revenue by restaurant:")
revenue = revenue_by_restaurant(orders)
for r in revenue:
    print(" ", r, ": Rs.", revenue[r])

item, qty = most_ordered_item(orders)
print("\nMost frequently ordered item:", item, "(", qty, "units )")

name, rev = top_restaurant(revenue)
print("\nTop-performing restaurant:", name, "with revenue Rs.", rev)
```
